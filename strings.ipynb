```python
#finding duplicate characters
def find_duplicates(text):
    counts = {}
    for ch in text:
        counts[ch] = counts.get(ch, 0) + 1
    return{ch: counts for n, n in counts.items() if n > 1}

print(find_duplicates("computer"))
```

```python
#find character frequency
def char_frequency(text):
    counts = {}
    for ch in text:
        counts[ch] = counts.get(ch, 0) + 1
    return{ch: n for ch, n in counts.items() if n > 1}
print(char_frequency("computer programming"))
```

```python
from collections import Counter
def are_anagrams(a, b):
    return Counter(a) == Counter(b)

print(are_anagrams("silent", "listen"))
print(are_anagrams("computer", "sushmita"))
```
