```python
#write a function that returns the season by taking the month as param.
Spring = ['January', 'February', 'March']
Summer = ['April', 'May', 'June']
Monsoon = ['July', 'August', 'September']
Winter = ['October', 'November', 'December']
def check_season(month):
    if month in Spring:
        print("Current season is Spring.")
    elif month in Summer:
        print("Current season is Summer.")
    elif month in Monsoon:
        print("Current season is Monsoon.")
    elif month in Winter:
        print("Current season is Winter.")
    else:
        print("Enter a valid input.")
month = input("Enter month: ")
check_season(month)

```

```python
#Area of circle
pie = 3.14
def area_of_circle(radius):
    if isinstance(radius, (int, float)) and radius >= 0:
        area = pie * radius ** 2
        print("Area of circle is:", area)
    else:
        print("Enter a valid input.")

radius = float(input("Enter radius: "))
area_of_circle(radius)
```

```python
#function that converts C to F.
def celsius_to_fahrenheit(celsius):
    if isinstance(celsius, (int, float)) and celsius >= 0:
        fahrenheit = (celsius * 9 / 5) + 32
        print("Fahrenheit is:", fahrenheit)
    else:
        print("Enter a valid input.")

celsius = float(input("Enter celsius: "))
celsius_to_fahrenheit(celsius)
```

```python
#function which returns the slope of a linear equation.
def calculate_slope(x1, y1, x2, y2):
    if x2 == x1:
        raise ValueError("Slope is undefined for vertical line.")
    elif y2 == y1:
        raise ValueError("Slope is undefined for horizontal line.")
    else:
        return (y2 - y1) / (x2 - x1)
print(calculate_slope(1, 2, 4, 11))
```

```python
import math
def solve_quadratic(a, b, c):
    if a == 0:
        raise ValueError("Not a quadratic equation: 'a' must not be 0.")
    discriminant = b**2 - 4*a*c
    if discriminant > 0:
        root1 = (-b + math.sqrt(discriminant))/(2*a)
        root2 = (-b - math.sqrt(discriminant))/(2*a)
        return root1, root2

    elif discriminant < 0:
        root = -b / (2*a)
        return root

    else:
        real_part = -b / (2*a)
        imaginary_part = math.sqrt(-discriminant) / (2*a)
        return complex(real_part, imaginary_part), complex(real_part, -imaginary_part)

print(solve_quadratic(1, -5, 6))
print(solve_quadratic(1, 2, 1))
```

```python
def remove_item(lst, item):
    if item in lst:
        return [x for x in lst if x != item]
    return lst
result = remove_item([1, 4, 9, 16], 4)
print(result)
```

```python
def sum_numbers(num):
    total = 0
    for i in range(1, num + 1):
        total += i
    return total
sum_numbers(10)
```

```python
def sum_even(num):
    even_sum = 0
    odd_sum = 0
    for i in range(1, num + 1):
        if i % 2 == 0:
            even_sum += i
        else:
            odd_sum += i
    return even_sum
sum_even(100)
```

```python
def sum_odd(num):
    odd_sum = 0
    even_sum = 0
    for i in range(1, num + 1):
        if i % 2 != 0:
            odd_sum += i
        else:
            even_sum += i
    return odd_sum
sum_odd(100)
```

```python
#calculate even and odd nums in a range of numbers
def even_num(n):
    evens = []
    odds = []
    for i in range(n + 1):
        if i % 2 == 0:
            evens.append(i)
        else:
            odds.append(i)
    return evens, odds
even_num(10)
```

```python
def evens_and_odds(n):
    evens = 0
    odds = 0
    for digit in range(n):
        if int(digit) % 2 == 0:
            evens += 1
        else:
            odds += 1
    return evens, odds
evens_and_odds(100)
```

```python
def calculate_factorial(num):
    factorial = 1
    for i in range(1, num + 1):
        factorial *= i
    return factorial
calculate_factorial(3)
```

```python
from collections import Counter
import math
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)

def calculate_median(numbers):
    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)
    mid = n // 2

    if n % 2 == 1:
        return sorted_numbers[mid]
    return(sorted_numbers[mid - 1] + sorted_numbers[mid]) / 2

def calculate_mode(numbers):
    counts = Counter(numbers)
    max_count = max(counts.values())
    modes = [num for num, count in counts.items() if count == max_count]

    return modes[0] if len(modes) == 1 else modes

def calculate_range(numbers):
    return max(numbers) - min(numbers)

def calculate_variance(numbers):
    mean = calculate_mean(numbers)
    return sum((x - mean) ** 2 for x in numbers) / len(numbers)

def calculate_std(numbers):
    return math.sqrt(calculate_variance(numbers))

data = [2, 4, 4, 4, 5, 5, 7, 9]

print(calculate_mean(data))
print(calculate_median(data))
print(calculate_mode(data))
print(calculate_range(data))
print(calculate_variance(data))
print(calculate_std(data))
```

```python
def is_prime(num):
    for i in range(2, int(num ** 0.5) + 1):
        if num < 2:
            print(num, "is not prime.")
        elif num % i == 0:
            print(num, "is not prime.")
        else:
            print(num, "is prime.")
is_prime(17)

```

```python
def all_unique(items):
    return len(items) == len(set(items))
print(all_unique([1, 2, 3, 4, 5]))
print(all_unique([1, 2, 2, 3, 5]))
print(all_unique(['a', 'b', 'a']))
```

```python
def all_unique(items):
    seen = set()
    for item in items:
        if item in seen:
            return False
        seen.add(item)
    return True
all_unique([10, 20, 30, 40, 50])
all_unique([10, 20, 20, 30, 40, 10])
```

```python
def same_data_type(items):
    if not items:
        return True

    first_type = type(items[0])
    return all(type(item) == first_type for item in items)
same_data_type([1, 2, 3, 4, 5])
same_data_type(['a', 'b', 'c', 'd', 'e'])
same_data_type([1.0, 2.0, 3, 4, 5.0])
```
