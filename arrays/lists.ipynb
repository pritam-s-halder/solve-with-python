```python
#find min value
numbers = [2, 4, 1, 6, 8, 9, 17, 12, 5]
for number in numbers:
    print(min(numbers))
    break
```

```python
#reverse an array
subjects = ['maths', 'computer', 'programming', 'engineering']
reverse_subjects = []
for i in range(len(subjects) - 1, -1, -1):
    reverse_subjects.append(subjects[i])
print(reverse_subjects)
```

```python
def second_largest(numbers):
    unique_numbers = sorted(set(numbers))

    if len(unique_numbers) == 1:
        return None

    return unique_numbers[-2]
print(second_largest([10, 20, 4, 78, 91, 5]))
print(second_largest([10, 20, 20, 7]))
print(second_largest([5, 5, 5]))
```

```python
def search_element(items, target):
    if target in items:
        return items.index(target)
    return -1
print(search_element([1, 2, 3, 4, 5], 5))
```

```python
def merge_arrays(items1, items2):
    return items1 + items2
print(merge_arrays([10, 30, 50], [20, 40, 60]))
```

```python
def second_largest(numbers):
    first = second = None
    for num in numbers:
        if first is None or num > first:
            second = first
            first = num
        elif num != first and (second is None or num > second):
            second = num
    return second
print(second_largest([10, 30, 50]))
print(second_largest([8, 7, 6]))
print(second_largest([5, 5, 5]))
```

```python
def is_armstrong(n):
    digits = str(n)
    power = len(digits)
    return n == sum(int(d) ** power for d in digits)
print(is_armstrong(153))
print(is_armstrong(9474))
print(is_armstrong(1223))
```

```python
y_intercept = lambda m, x, y: y - m * x
slope = lambda x1, y1, x2, y2: (y2 - y1) / (x2 - x1)

m = slope(1, 5, 3, 11)
b = y_intercept(m, 1, 5)
print(f"y = {m}x + {b}")
```
