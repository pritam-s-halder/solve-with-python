```python
#count word frequencies
from collections import Counter
text = input("Enter a sentence: ")
words = text.lower().split()
word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print(word_count)
```

```python
#character count frequency
from collections import Counter
text = input("Enter a sentence: ")
letters = text.lower().replace(" ", "")
letter_count = {}
for letter in letters:
    if letter in letter_count:
        letter_count[letter] += 1
    else:
        letter_count[letter] = 1
print(letter_count)
```

```python
#duplicate frequency counter
items = ['cheese', 'milk', 'butter', 'cheese', 'milk']
frequency = {}
for item in items:
    frequency[item] = frequency.get(item, 0) + 1
duplicates = [item for item, count in frequency.items() if count > 1]
print("Frequency:", frequency)
print("Duplicates:", duplicates)
```

```python
#find unique items
items = [1, 2, 2, 3, 4, 4, 5]
frequency = {}
for item in items:
    frequency[item] = frequency.get(item, 0) + 1
unique_items = []
for item, count in frequency.items():
    if count == 1:
        unique_items.append(item)
print("Frequency:", frequency)
print("Unique items:", unique_items)
```

```python
#find highest-value entry
scores = {
    "Aarav": 86,
    "Diya": 98,
    "Rohan": 78,
}
highest_key = max(scores, key=scores.get)
print(highest_key)
print(scores[highest_key])
```

```python
#merge two key-value collections
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

merged = {}

for key, value in dict1.items():
    merged[key] = [value]
for key, value in dict2.items():
    if key in merged:
        merged[key].append(value)
    else:
        merged[key] = [value]
print(merged)
```
