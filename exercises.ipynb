```python
import re

text = """the position of some particles on the horizontal x-axis are -12, -4, -3 and -1 in the negative direction, 0 at origin, 4 and 8 in the positive direction."""

positions = [int(x) for x in re.findall(r"-?\d+", text)]

print(positions)

distance = max(positions) - min(positions)
print(distance)
```

```python
import re
def far_distance(text):
    positions = [int(num) for num in re.findall(r"-?\d+", text)]
    if len(positions) < 2:
        return 0
    else:
        return max(positions) - min(positions)

text = """The position of some particles on the horizontal x-axis are -12, -4, -3 and -1 in the negative direction, 0 at origin, 4 and 8 in the positive direction. Extract these numbers from this whole text and find the distance between the two furthest particles."""

print(far_distance(text))
```

```python
import re
from collections import Counter

def most_frequent(text, top_n=5):
    words = re.findall(r"\b[a-zA-Z]+\b", text.lower())
    word_counts = Counter(words)
    return word_counts.most_common(top_n)

sentence = '''%I $am@% a %tea@cher%, &and& I lo%#ve %tea@ching%;. There $is nothing; &as& mo@re rewarding as educa@ting &and& @emp%o@wering peo@ple. ;I found tea@ching m%o@re interesting tha@n any other %jo@bs. %Do@es thi%s mo@tivate yo@u to be a tea@cher!?'''

print(most_frequent(sentence))
```
