```python
#check palindrome
text = input("Enter a word: ")

if text == text[::-1]:
    print(text, "is a palindrome.")

else:
    print(text, "is not a palindrome.")
```

```python
#palindrome that ignores capital letters and spaces
text = input("Enter a word or sentence: ")
clean_text = text.lower().replace(" ", "")

if clean_text == clean_text[::-1]:
    print(text, "is a palindrome.")
else:
    print(text, "is not a palindrome.")
```

```python
#to count digits but with math approach
import math

def count_digits(n):
    if n == 0:
        return 1
    return math.floor(math.log(abs(n))) + 1

print(count_digits(-54638574))
print(count_digits(0))
```

```python
n = int(input("Enter a number: "))

if n < 2:
    print(n, "is not a prime number.")
else:
    is_prime = True

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        print(n, "is a prime number.")
    else:
        print(n, "is not a prime number.")
```

```python
n = int(input("Enter N: "))
for num in range(2, n + 1):
    is_prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num)
```

```python
#generate a fibonacci series
n = int(input("Enter no. of terms: "))
a, b = 0, 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
#generate a fibonacci series in list
n = int(input("Enter N: "))
a, b = 0, 1
fib = []
for i in range(n):
    fib.append(a)
    a, b = b, a + b
print(fib)
```

```python
#generate fibonacci series with use of while loop
n = int(input("Enter N: "))
a, b = 0, 1

count = 0

while count < n:
    print(a, end=" ")
    a, b = b, a + b
    count += 1
```

```python
#The sum of all evens is 2550. And the sum of all odds is 2500.
n = int(input("Enter N: "))
even_sum = 0
odd_sum = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i

print("Even numbers: ", even_sum)
print("Odd numbers: ", odd_sum)
```
